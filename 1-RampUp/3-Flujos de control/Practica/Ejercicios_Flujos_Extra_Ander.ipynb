## Ejercicio Extra
Programa una calculadora que

* Pregunte qué operación quieres realizar (multiplicación/división/suma/resta/potencia/salir)
* Pida dos números 
* Realice la operación
* Siga funcionando hasta que el usuario elija salir

#### Bonus:
* Que guarde todas las operaciones realizadas y las imprima por pantalla cuando el usuario elija esa opción
* Que sea robusta frente a entradas erróneas del usuario o frente a operaciones como dividir por 0

```python
import math
import pprint
import time
```

```python
operacion = 0
salir = False
try:
    while operacion != 6 and not salir:

        operacion = int(input("Introduce la operación a realizar (1. Multiplicación, \n" \
        "2. División\n" \
        "3. Suma\n" \
        "4. Resta\n" \
        "5. Potencia\n" \
        "6. Salir): "))
    
        if operacion >0 and operacion <6:
            numero1 = float(input("Introduce el primer número: "))
            numero2 = float(input("Introduce el segundo número: "))

            match operacion:
                case 1:
                    resultado = numero1 * numero2
                    print(f"El resultado de la multiplicación {numero1} * {numero2} es: {resultado}")
                case 2:
                    resultado = numero1 / numero2
                    print(f"El resultado de la división {numero1} / {numero2} es: {resultado}")
                case 3:
                    resultado = numero1 + numero2
                    print(f"El resultado de la suma {numero1} + {numero2} es: {resultado}")
                case 4:
                    resultado = numero1 - numero2
                    print(f"El resultado de la resta {numero1} - {numero2} es: {resultado}")
                case 5:
                    resultado = numero1 ** numero2
                    print(f"El resultado de la potencia {numero1} ** {numero2} es: {resultado}")
                case 6:
                    salir = True

except ValueError:
    print("Debes introducir un número para la operación.")
except ZeroDivisionError:
    print("No se puede dividir entre cero.")
```
