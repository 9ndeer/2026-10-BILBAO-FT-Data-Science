## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import random

palabras = ["sistema", "python", "futbol", "entrenador", "bilbao"]
palabra = random.choice(palabras)

aciertos = []
usadas = []
errores = 0

while errores < 6:
    progreso = ""
    for i in palabra:
        if i in aciertos:
            progreso += i + " "
        else:
            progreso += "_ "
    print(progreso)

    if "_" not in progreso:
        print("¡Ganaste!")
        break

    letra = input("Letra: ").lower()

    if letra in usadas:
        print("Esa ya la probaste")
        continue

    usadas.append(letra)

    if letra in palabra:
        aciertos.append(letra)
    else:
        errores += 1
        print(f"Fallo {errores}/6")

if errores == 6:
    print("Perdiste, la palabra era:", palabra)
````
