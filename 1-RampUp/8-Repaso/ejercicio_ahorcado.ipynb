## Juego del Ahorcado
El funcionamiento es que vaya mostrando por pantalla ```_ _ _ _ _``` las rayas bajas correspondientes al número de letras que tiene la palabra a adivinar. El usuario deberá insertar una letra y, si esa letra existe, se mostrará o mostrarán las letras acertadas ```S _ S _ _```. 

Si falla, se mostrará el número de errores que lleva y, si lleva 6 errores, entonces se mostrará "Fin del juego".
Encapsúlalo en una función.

````python
import random

MAX_ERRORES = 6


def ahorcado(palabra=None):
    if palabra is None:
        palabra = random.choice(["sistema", "python", "futbol", "entrenador", "bilbao"])
    palabra = palabra.lower()

    aciertos = set()
    usadas = set()
    errores = 0

    while errores < MAX_ERRORES:
        progreso = " ".join(l if l in aciertos else "_" for l in palabra)
        print(f"\n{progreso}")

        if all(l in aciertos for l in palabra):
            print("¡Ganaste!")
            return True

        letra = input("Introduce una letra: ").lower().strip()

        if len(letra) != 1 or not letra.isalpha():
            print("Solo una letra, porfa.")
            continue
        if letra in usadas:
            print("Esa ya la has probado.")
            continue

        usadas.add(letra)

        if letra in palabra:
            aciertos.add(letra)
        else:
            errores += 1
            print(f"Fallo. Errores: {errores}/{MAX_ERRORES}")

    print(f"\nFin del juego. La palabra era: {palabra}")
    return False


if __name__ == "__main__":
    ahorcado()
````
